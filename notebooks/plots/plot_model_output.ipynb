```python
from mpl_toolkits.axes_grid1 import make_axes_locatable

def plot_map(data, folder_path, title, file_name, vmin=0, vmax=1, cmap='viridis'):
    lats = np.linspace(-77, 90, 167)
    lons = np.linspace(-180, 180, 360)
    lon_grid, lat_grid = np.meshgrid(lons, lats)

    fig = plt.figure(figsize=(20, 14))
    ax = plt.axes(projection=ccrs.PlateCarree())
    ax.set_global()
    ax.set_title(title, fontsize=30, loc='center')

    pcm = ax.pcolormesh(lon_grid, lat_grid, data,
                        transform=ccrs.PlateCarree(),
                        shading='auto', cmap=cmap,
                        vmin=vmin, vmax=vmax)

    # Set land color to black
    land = cfeature.NaturalEarthFeature(
        'physical', 'land', '110m',
        edgecolor='face', facecolor='black'
    )
    ax.add_feature(land)

    # Attach colorbar with the same height as the plot
    divider = make_axes_locatable(ax)
    cax = divider.append_axes("right", size="5%", pad=0.4, axes_class=plt.Axes)  # pad controls spacing
    cbar = plt.colorbar(pcm, cax=cax, orientation='vertical')
    cbar.set_label('CO₂ Flux (mol/m²/s)', fontsize=30, labelpad=20)  # label font size
    cbar.ax.tick_params(labelsize=30)  # tick font size


    plt.tight_layout()

    timestamp = datetime.now().strftime('%Y-%m-%d_%H-%M')
    path = folder_path + '/' + timestamp + '_' + file_name + 'plot.png'
    plt.savefig(path, format='png', dpi=300, bbox_inches='tight')
    plt.close()
```

```python
start_year = 2018
end_year = 2018
dataset_id = "exp1"
unet_path = "../../outputs/u-net/best"


targets, pred = get_masked_prediction_map(unet_path, dataset_id, start_year=start_year, end_year=end_year, target_index=3)

plot_map(pred[0],unet_path,"UNet CO2 Flux-pre prediction January 2018","pred_jan_18",vmin=-5,vmax=5,cmap="RdBu_r")
```
