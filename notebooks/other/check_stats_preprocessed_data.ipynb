```python
import os
import numpy as np
from utils.data_loader import load_as_maps, load_for_mlp
from utils.ocean_basins import get_combined_full_mask
```

```python
data_path = "../../data/"
```

```python
datasets = ["exp8"]#,"exp3","exp5","exp6","exp7","exp8"]

feature_names = ['SST', 'SAL', 'ice_frac', 'mixed_layer_depth', 'heat_flux_down', 'water_flux_up', 'stress_X', 'stress_Y', 'currents_X', 'currents_Y']
regions = ["NORTH_ATLANTIC", "SOUTHERN_OCEAN", "EQ_PACIFIC", "ARCTIC"]

for dataset in datasets:
    print(dataset)
    _, targets = load_as_maps(datasets=[dataset], target_index=3)


    for region in regions:
        print(region)
        mask = get_combined_full_mask(region)
        region_targets = targets[:,mask]


        abs_mean = np.mean(np.abs(region_targets))
        mean = np.mean(region_targets)
        std = np.std(region_targets)

        print(f"CO2 Flux Pre: Abs Mean: {abs_mean:.4f}, Mean: {mean:.4f}, Std: {std:.4f}")
```

```python
datasets = ["exp1"]#,"exp3","exp5","exp6","exp7","exp8"]

feature_names = ['SST', 'SAL', 'ice_frac', 'mixed_layer_depth', 'heat_flux_down', 'water_flux_up', 'stress_X', 'stress_Y', 'currents_X', 'currents_Y']

for dataset in datasets:
    print(dataset)
    features, targets = load_for_mlp(datasets=[dataset], target_index=3)
    
    mask = features[:, 10] == 1
    features = np.delete(features, [10], axis=1)
    features = features[mask]
    targets = targets[mask]

    for i in range(len(feature_names)):
        mean = np.mean(features[:,i])
        std = np.std(features[:,i])
        min_val = np.min(features[:,i])
        max_val = np.max(features[:,i])

        print(f"{feature_names[i]}: Mean: {mean:.4f}, Std: {std:.4f}, Min: {min_val:.4f}, Max: {max_val:.4f}")

    mean = np.mean(targets)
    std = np.std(targets)
    min_val = np.min(targets)
    max_val = np.max(targets)

    print(f"CO2 Flux Pre: Mean: {mean:.4f}, Std: {std:.4f}, Min: {min_val:.4f}, Max: {max_val:.4f}")
```
