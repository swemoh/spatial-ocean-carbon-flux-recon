```python
import numpy as np
import matplotlib.pyplot as plt

# Define longitude range (degrees)
lon = np.linspace(-180, 180, 1000)

# Sine encoding (convert to radians first)
lon_sin = np.sin(np.deg2rad(lon))
lon_cos = np.cos(np.deg2rad(lon))

plt.figure(figsize=(8, 4))
plt.plot(lon, lon_sin, label='sin(λ)')
plt.plot(lon, lon_cos, label='cos(λ)')
plt.title('Sine and Cosine Encoding of Longitude')
plt.xlabel('Longitude (°)')
plt.ylabel('Encoded value')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# Define longitude range (degrees)
month = np.linspace(1, 12, 12)

# Sine encoding (convert to radians first)
month_sin = np.sin(2 * np.pi * month / 12)
month_cos = np.cos(2 * np.pi * month / 12)

plt.figure(figsize=(8, 4))
plt.plot(month, month_sin, label='month_sin')
plt.plot(month, month_cos, label='month_cos')
plt.title('Sine and Cosine Encoding of Month')
plt.xlabel('Longitude (°)')
plt.ylabel('Encoded value')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
